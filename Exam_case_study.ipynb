```python
# CASE STUDY: UNLOCKING REVENUE
## Identifying Bottlenecks in Regional Sales Pipeline

*Project Type:* Business Case-Based Practical
*Role:* Freelance Business Analyst
*Tools Used:* MySQL, Microsoft Excel / Power BI
*Dataset:* RegionalSales2025.csv
*Author:* Angel Bareja

---

## 1. Executive Summary

This case study focuses on analyzing sales transactions for a retail chain experiencing inconsistent sales performance across different regions. The organization wants to understand why certain regions underperform, which products contribute to revenue losses, and how sales operations can be improved.

Using SQL, the transaction dataset will be analyzed to identify monthly sales trends, order cancellation and return rates, revenue at risk, product category performance, sales agent productivity, and customers with frequent returns.

The analysis will provide data-driven insights and recommendations to improve sales performance, reduce operational bottlenecks, and identify high-value business opportunities.

Note: Actual findings and numerical results must be added after executing the SQL queries on the supplied dataset.

## 2. Business Problem

The retail chain faces several business challenges:

- Inconsistent sales across its East, West, North, and South regions.
- Revenue at risk because of cancelled and returned orders.
- Limited visibility into the performance of individual products and categories.
- Difficulty identifying the highest-performing sales agents.
- Repeated returns from certain customers.
- Lack of a consolidated view of monthly sales performance.

The company requires a structured analysis to identify the causes of these issues and recommend practical improvements.

## 3. Project Objectives

The main objectives are:

1. Analyze monthly sales trends across all regions.
2. Calculate the percentage of cancelled and returned orders by region.
3. Identify the three regions and products with the highest revenue at risk.
4. Calculate average order value by product category.
5. Identify the top five sales agents based on completed revenue.
6. Measure each product category's contribution to total completed revenue.
7. Identify customers with three or more returned orders.
8. Develop actionable recommendations to improve sales performance.

## 4. Dataset Description

The dataset, RegionalSales2025.csv, contains retail order transaction information.

| Column | Description |
|---|---|
| OrderID | Unique order identifier |
| Date | Transaction date |
| CustomerID | Unique customer identifier |
| Region | Sales region |
| ProductName | Name of the product |
| Category | Product category |
| Quantity | Number of units sold |
| UnitPrice | Selling price per unit |
| TotalAmount | Total transaction amount |
| OrderStatus | Completed, Cancelled, or Returned |
| SalesAgent | Sales agent assigned to the order |

*Key assumptions:* Each row represents one order, order statuses are standardized, and the revenue analysis uses TotalAmount. These assumptions should be verified against the actual dataset before presenting final results.

## 5. Tools and Technologies

*MySQL:* Used to create the database, import transactions, validate records, aggregate sales, and perform business analysis.

*Microsoft Excel:* Can be used to create pivot tables, charts, and summary reports.

*Power BI:* Can be used to build an interactive dashboard with regional filters, monthly trends, category comparisons, and sales performance KPIs.

## 6. Methodology

### Step 1: Data Collection

Import RegionalSales2025.csv into a MySQL database named RegionalSalesDB.

### Step 2: Data Cleaning

- Check for duplicate order IDs.
- Identify missing customer, region, product, and sales agent values.
- Validate dates, quantities, prices, and total amounts.
- Standardize order statuses.
- Investigate inconsistent transaction amounts before correcting them.

### Step 3: SQL Analysis

Execute SQL queries to analyze sales trends, order statuses, revenue at risk, category performance, sales agent productivity, and customer returns.

### Step 4: Data Visualization

Export query results to Excel or connect the database to Power BI. Create charts and tables to make the results easier to understand.

### Step 5: Business Recommendations

Use the analysis to identify underperforming regions, operational issues, high-value products, and opportunities to improve revenue.

## 7. Business Analysis and Expected Findings

### Analysis 1: Monthly Sales Trend

*SQL objective:* Calculate completed revenue for each month and region.

*Expected business value:* Identify seasonal patterns, declining sales periods, and regions that need additional marketing or sales support.

*Recommended visualization:* Line chart showing monthly completed revenue, with separate lines for each region.

### Analysis 2: Cancellation and Return Rates

*SQL objective:* Calculate cancelled and returned order percentages for each region.

*Expected business value:* Identify locations with higher order failure rates and investigate possible delivery, product quality, customer service, or order processing issues.

*Recommended visualization:* Clustered bar chart comparing cancellation and return rates by region.

### Analysis 3: Revenue at Risk

*SQL objective:* Rank regions and products by the value of cancelled and returned orders.

*Expected business value:* Prioritize products and regions that require operational investigation.

*Recommended visualization:* Bar chart showing the top three products and regions by revenue at risk.

*Important distinction:* The amount associated with a cancelled or returned order is a proxy for revenue at risk. It does not automatically equal realized financial loss because refunds, restocking, recoveries, and accounting adjustments may affect the actual result.

### Analysis 4: Average Order Value by Category

*SQL objective:* Calculate average completed order amount for each product category.

*Expected business value:* Identify categories generating higher-value orders and opportunities for cross-selling or product bundling.

*Recommended visualization:* Bar chart comparing average order value across categories.

### Analysis 5: Sales Agent Performance

*SQL objective:* Rank the top five sales agents by completed revenue.

*Expected business value:* Recognize high-performing agents and identify practices that may be useful for training other team members.

*Recommended visualization:* Horizontal bar chart showing completed revenue by sales agent.

###
 Analysis 6: Category-Wise Revenue Contribution

*SQL objective:* Calculate total completed revenue and each category's percentage contribution to the grand total.

*Expected business value:* Understand which categories drive the largest share of revenue and which may require pricing, promotion, or assortment changes.

*Recommended visualization:* Bar chart or donut chart showing category revenue contribution.

### Analysis 7: Frequent Customer Returns

*SQL objective:* Identify customers with at least three returned orders.

*Expected business value:* Investigate recurring return patterns while avoiding unsupported assumptions about customer behavior.

*Recommended visualization:* Bar chart showing the number of returned orders by customer ID.

## 8. Key Performance Indicators (KPIs)

The following KPIs should be calculated from the dataset:

| KPI | Calculation |
|---|---|
| Total Orders | Count of all orders |
| Completed Revenue | Sum of amounts for completed orders |
| Completion Rate | Completed orders ÷ total orders × 100 |
| Cancellation Rate | Cancelled orders ÷ total orders × 100 |
| Return Rate | Returned orders ÷ total orders × 100 |
| Average Order Value | Completed revenue ÷ completed orders |
| Revenue at Risk | Amount associated with cancelled and returned orders |
| Category Contribution | Category completed revenue ÷ total completed revenue × 100 |
| Top Sales Agent | Agent with the highest completed revenue |

For consistency, order-based metrics should count orders, while revenue metrics should sum the transaction amounts. The exact definitions should be documented in the final dashboard.

## 9. Business Recommendations

Based on the results obtained from the SQL analysis, the following actions can be considered:

1. *Improve regional performance:* Develop targeted sales and marketing strategies for regions with declining completed revenue.
2. *Reduce cancellations:* Investigate order processing delays, stock availability, payment failures, and delivery issues in regions with high cancellation rates.
3. *Control product returns:* Review product quality, product descriptions, packaging, and customer feedback for products with frequent returns.
4. *Optimize product categories:* Give additional attention to categories with strong revenue contribution while investigating categories with weak performance.
5. *Improve sales productivity:* Study the working methods of top-performing sales agents and use the findings to improve team training.
6. *Monitor customer return patterns:* Review repeated returns using appropriate customer service procedures and return-policy checks.
7. *Use regular reporting:* Build a monthly dashboard to monitor sales, cancellations, returns, and revenue performance.

These recommendations should be prioritized according to the actual results rather than assumed findings.

## 10. Expected Project Deliverables

The completed project should include:

- A MySQL database containing the imported sales dataset.
- SQL scripts for all seven required analyses.
- Exported query results in CSV or Excel format.
- An Excel report or Power BI dashboard.
- Charts showing sales trends and performance comparisons.
- A case study documenting the business problem, methodology, findings, and recommendations.

## 11. Conclusion

This case study demonstrates how SQL can be used to transform retail transaction data into meaningful business insights. By analyzing monthly sales, regional performance, cancellations, returns, category contribution, sales agent productivity, and customer return patterns, a business analyst can help the retail chain identify operational bottlenecks and prioritize improvement opportunities.

Combining SQL analysis with Excel or Power BI visualization provides a practical foundation for evidence-based decision-making.

The final conclusion should be updated with actual figures, charts, and findings after the dataset has been analyzed.

---

*Project Author:* Angel Bareja
*Project Title:* Unlocking Revenue — Identifying Bottlenecks in Regional Sales Pipeline
*Technologies:* MySQL | Excel / Power BI | Data Analysis
```
